```python
import pandas as pd

race_results = pd.read_csv("../data/raw/race_results/british_gp_2025.csv")
q_results = pd.read_csv("../data/raw/qualifying/british_gp_2025.csv")
laps = pd.read_csv("../data/raw/laps/british_gp_2025.csv")
stints = pd.read_csv("../data/raw/stints/british_gp_2025.csv")
pit_stops = pd.read_csv("../data/raw/pit_stops/british_gp_2025.csv")
weather = pd.read_csv("../data/raw/weather/british_gp_2025.csv")

print("All six datasets loaded successfully!")
```

```python
datasets = {
    "RACE RESULTS": race_results,
    "QUALIFYING": q_results,
    "LAPS": laps,
    "STINTS": stints,
    "PIT STOPS": pit_stops,
    "WEATHER": weather
}

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)
    print(df.columns.tolist())
```

```python
from pathlib import Path
import pandas as pd

race_folder = Path("../data/raw/race_results")

for year in range(2020, 2026):
    files = list(
        race_folder.glob(f"{year}_*_grand_prix.csv")
    )

    non_empty = 0
    empty = 0

    for file in files:
        df = pd.read_csv(file)

        if len(df) > 0:
            non_empty += 1
        else:
            empty += 1

    print(
        f"{year}: {non_empty} non-empty | "
        f"{empty} empty"
    )
```
