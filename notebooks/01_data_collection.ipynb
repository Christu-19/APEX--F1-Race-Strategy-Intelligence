```python
import fastf1
import pandas as pd
import numpy as np
```

```python
import os

os.makedirs("../data/fastf1_cache", exist_ok=True)

fastf1.Cache.enable_cache("../data/fastf1_cache")

print("FastF1 cache enabled successfully!")
```

```python
session = fastf1.get_session(2025,"British","R")
session.load()
```

```python
results = session.results
results.head()
```

```python
results.to_csv("../data/raw/race_results/british_gp_2025.csv",index=False)
```

```python
qualifying = fastf1.get_session(2025, "British", "Q")
qualifying.load()
```

```python
q_results = qualifying.results
q_results.head()
```

```python
import os

print(os.getcwd())
```

```python
q_results.to_csv(
    "../data/raw/qualifying/british_gp_2025.csv",
    index=False
)

print("Successful!")
```

```python
laps = session.laps
laps.head()
```

```python
laps.to_csv("../data/raw/laps/british_gp_2025.csv",index=False)
print("succesful")
```

```python
laps.columns.tolist()
```

```python
stint_col = ["Driver","LapNumber","LapTime","Compound","TyreLife","Stint"]
stints = laps[stint_col].copy()
```

```python
stints.head()
```

```python
stints.to_csv("../data/raw/stints/british_gp_2025.csv",index=False)
print("succesful")
```

```python
print(laps.columns.tolist())
```

```python
pit_stops = laps[laps["PitInTime"].notna()][["Driver", "LapNumber", "PitInTime", "PitOutTime"]].copy()
```

```python
pit_stops.head()
```

```python
pit_stops.to_csv("../data/raw/pit_stops/british_gp_2025.csv",index=False)
print("Succesful")
```
