```python
import pandas as pd

race_results = pd.read_csv("../data/raw/race_results/british_gp_2025.csv")

q_results = pd.read_csv("../data/raw/qualifying/british_gp_2025.csv")

laps = pd.read_csv("../data/raw/laps/british_gp_2025.csv")

stints = pd.read_csv("../data/raw/stints/british_gp_2025.csv")

pit_stops = pd.read_csv("../data/raw/pit_stops/british_gp_2025.csv")

weather = pd.read_csv("../data/raw/weather/british_gp_2025.csv")

print("All six datasets loaded successfully!")
```

```python
print("RACE RESULTS :", race_results.shape)
print("QUALIFYING   :", q_results.shape)
print("LAPS         :", laps.shape)
print("STINTS       :", stints.shape)
print("PIT STOPS    :", pit_stops.shape)
print("WEATHER      :", weather.shape)
```

```python
dataset ={
    "RACE RESULTS": race_results,
    "QUALIFYING":q_results,
    "LAPS":laps,
    "STINTS":stints,
    "PIT STOPS":pit_stops,
    "WEATHER":weather
}

for name,df in dataset.items():
    print("\n"+"="*60)
    print(name)
    print("="*60)
    print(df.columns.tolist())
```
