```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path("../data/raw")

folders = [
    "race_results",
    "qualifying",
    "laps",
    "stints",
    "pit_stops",
    "weather"
]

years = range(2020, 2026)

for year in years:

    print("\n" + "=" * 60)
    print(f"SEASON {year}")
    print("=" * 60)

    # Find all race
    race_files = sorted(
        (BASE_DIR / "race_results").glob(f"{year}_*.csv")
    )

    print(f"Race results found: {len(race_files)}")

    for race_file in race_files:

        race_name = race_file.stem

        print(f"\n{race_name}")

        for folder in folders:

            file_path = BASE_DIR / folder / f"{race_name}.csv"

            if file_path.exists():
                print(f"  ✓ {folder}")
            else:
                print(f"  ✗ {folder}")
```
