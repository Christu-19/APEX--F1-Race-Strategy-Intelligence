```python
import fastf1
from pathlib import Path

BASE_DIR = Path("../data/raw/race_results")

years = range(2020, 2026)

for year in years:

    print("\n" + "=" * 60)
    print(f"SEASON {year}")
    print("=" * 60)

    # Get official season schedule
    schedule = fastf1.get_event_schedule(year)

    # Keep actual race rounds only
    schedule = schedule[schedule["RoundNumber"] > 0]

    # Existing race-result files
    existing_files = {
        file.stem
        for file in BASE_DIR.glob(f"{year}_*.csv")
    }

    print(f"Official races: {len(schedule)}")
    print(f"Existing files: {len(existing_files)}")

    print("\nMissing races:")

    missing_count = 0

    for _, event in schedule.iterrows():

        round_number = int(event["RoundNumber"])
        event_name = event["EventName"]

        expected_filename = (
            f"{year}_"
            f"{event_name.lower().replace(' ', '_')}.csv"
        )

        if expected_filename[:-4] not in existing_files:
            print(
                f"  ✗ Round {round_number}: "
                f"{event_name}"
            )
            missing_count += 1

    if missing_count == 0:
        print("  ✓ No missing races")

    print(f"\nTotal missing: {missing_count}")
```
